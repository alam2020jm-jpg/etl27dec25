```python
spark=SparkSession.builder.appName("MarketData").getOrCreate()
dfMarket=spark.read.options(header="true",inferSchema="true").csv("zafcon_trading_Market_corrected.csv")
```

```python
dfMarket.show(1)
```

```python
dfMarket=dfMarket.withColumn("TickSpread",(col("BestAsk")-col("BestBid")/(0.05))).drop("S.No")
```

Max and Average TickSpread required for comparision

```python
MaxTickSpread=dfMarket.agg(max("TickSpread")).first()[0]
AverageTickSpread=dfMarket.agg(avg("TickSpread")).first()[0]
print(f"MaxTickSpread:- {MaxTickSpread}")
print(f"AverageTickSpread:- {AverageTickSpread}")
```
